# Exercício 1 - Saudação Personalizada

  Elabore um programa que solicite ao usuário o seu nome e, em seguida, exiba uma mensagem de saudação personalizada, fazendo uso do nome informado.

**Orientações:**

  Utilize a função input() para capturar o nome;
  Exiba uma saudação no formato: "Olá, [nome]! É um prazer conhecê-lo(a)."

```python
nome = input("Insira seu nome: ")
print("Olá, ", "! É um prazer conhecê-lo(a).", sep = nome)
```
