# Projeto 002: Extrator e Consolidador de Logs

## Objetivo:
  Ler o log de um servidor em formato de texto, ignorar tudo o que for `INFO` ou `WARING`, capturar apenas linhas de `ERROR`, e gerar um relatório limpo mostrando apenas a data/hora e a descrição do erro.

## Requisitos do código:
  1. **Separação**: Transformar a string em uma lista de linhas;
  2. **Filtragem**: Percorrer cada linha da lista e verificar se contém a palavra `[ERROR]`. Se não contiver erro, simplesmente ignorar a linha;
  3. **Extração**: Para as linhas que possuem erro:
      * A Data e Hora;
      * A mensagem de erro;
  4. **Relatório**: Imprimir no console de forma amigável no seguinte formato:
    `Erro em 2026-08-13 14:28:45: Falha de autenticação no endpoit /login`

## Log para teste:
`log_servidor = """
2026-08-13 14:22:10 [INFO] Servidor iniciado com sucesso.
2026-08-13 14:23:05 [INFO] Conexão com banco de dados estabelecida.
2026-08-13 14:25:11 [WARNING] Uso de memória acima de 75%.
2026-08-13 14:28:45 [ERROR] Falha de autenticação no endpoint /login.
2026-08-13 14:30:00 [INFO] Rotina de backup iniciada.
2026-08-13 14:32:15 [ERROR] Timeout ao conectar na API de pagamento.
2026-08-13 14:35:01 [WARNING] Tentativas de login suspeitas detectadas.
2026-08-13 14:40:22 [ERROR] Banco de dados inacessível. Conexão recusada.
"""`

```python
# ENTRADA
log_servidor = """
2026-08-13 14:22:10 [INFO] Servidor iniciado com sucesso.
2026-08-13 14:23:05 [INFO] Conexão com banco de dados estabelecida.
2026-08-13 14:25:11 [WARNING] Uso de memória acima de 75%.
2026-08-13 14:28:45 [ERROR] Falha de autenticação no endpoint /login.
2026-08-13 14:30:00 [INFO] Rotina de backup iniciada.
2026-08-13 14:32:15 [ERROR] Timeout ao conectar na API de pagamento.
2026-08-13 14:35:01 [WARNING] Tentativas de login suspeitas detectadas.
2026-08-13 14:40:22 [ERROR] Banco de dados inacessível. Conexão recusada.
"""

# PROCESSAMENTO

  ## Separação
log_splitted = log_servidor.splitlines()

  ## Filtragem e extração
error_log = []
for line in log_splitted:
  if "[ERROR]" in line:
    date_time = line[:19] # armazenamento de data/hora
    error_tag = line.find("[ERROR]") # encontrando e armazenando o tag de erro .find()
    message = line[error_tag + len("[ERROR]"):]
    error_log.append((date_time, message.strip())) # armazenamento da tupla .append(), removendo espaços em branco extras .strip()

# SAÍDA
print("\n--- Relatório de Erros ---\n")
for date_time, message in error_log:
  print(f"Erro em {date_time}:{message}")
print("\n--- Fim do Relatório ---\n")
```
