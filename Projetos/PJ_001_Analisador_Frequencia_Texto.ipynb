# Projeto 001: Analisador de Frequência de Texto

## Objetivo:
  Receber um texto longo, limpá-lo (removendo pontuações e diferenças de maiúsculas/minúsculas) e contar quantas vezes cada palavra aparece no texto.

## Requisítos do código:
  1. **Padronização**: Converter todo o texto para letras minúsculas;
  2. **Limpeza**: Remover as pontuações básicas do texto;
  3. **Quebra de texto**: Transformar a string única em uma lista de palavras;
  4. **Contagem**: Criar um dicionário vazio. Percorrer a lista de palavras e adicionar cada nova palavra ao dicionário, contando suas aparições;
  5. **Saída**: Imprimir o dicionário final para verificar o resultado.

## Texto base para teste:

```python
txt_base = input("Insira o texto aqui: ")

txt_low = txt_base.lower() # TEXTO -> texto
txt_replaced = txt_low.replace(".", " ").replace(",", " ").replace(":", " ").replace(";", " ").replace("!", " ").replace("?", " ") # Remoção de pontuação
txt_splitted = txt_replaced.split() # Listagem da string

# Contador de repetição das palavras
count = {}
for word in txt_splitted:
  count[word] = count.get(word, 0) + 1

# SAÍDA
print("\n--- Relatório ---\n")
for word, word_count in count.items():
  print(f"{word}: {word_count} vez(es)")
print("\n--- Fim do Relatório ---")
```
