# Exercício 2 - Calculadora das Quatro Operações

Desenvolva uma calculadora simples que receba dois números fornecidos pelo usuário e exiba o resultado das quatro operações aritméticas básicas: adição, subtração, multiplicação e divisão.

**Orientações:**
* Considere que os números podem ser inteiros ou decimais — utilize a conversão adequada (int() ou float());

* Exiba cada operação acompanhada de seu respectivo resultado, de forma clara e organizada;

* Atente para a possibilidade de divisão por zero.

```python
try:

  num1 = float(input('Digite o primeiro número:'))
  num2 = float(input('Digite o segundo número:'))

  print('Adição:', num1 + num2)
  print('Subtração:', num1 - num2)
  print('Multiplicação:', num1 * num2)
  print('Divisão:', num1 / num2)

except ZeroDivisionError:
  print('Erro: Não é possível dividir por zero')
except ValueError:
  print('Erro: Insira um valor válido')
```
