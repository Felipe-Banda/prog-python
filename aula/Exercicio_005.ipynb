# Exercício 5 - Analisador de Números com Loops e Operações Lógicas

**Enunciado:**
Crie um programa que analise uma lista de números inteiros digitados pelo usuário. O programa deve:
* Pedir ao usuário para digitar números até que ele digite "fim" (use um loop while);
* Guardar os números em uma lista (converta com int());
* Usando break e continue:
  * Com continue - pule os números negativos ao somar;
  * Com break - pare o programa se o usuário digitar o número 0.
* Calcule e imprima:
  * A soma dos números positivos (usando continue para pular os negativos)
  * A quantidade de números pares e ímpares (usando o operador %);
  * A média dos números (use sum() e len());
* Use range() com 3 argumentos para imprimir os números da lista em ordem decrescente (do último para o primeiro).

```python
# INICIALIZAÇÃO
int_list = []
fim = False
soma_pos, cont_par, cont_impar, media = 0, 0, 0, 0
linha = '\n---\n'

# LOOP ENTRADA + .APPEND()
while fim == False:
  entrada = input('Digite um número inteiro ou "fim" para finalizar a lista: ')
  entrada_clean = entrada.lower().strip()
  if entrada_clean == 'fim':
    fim = True
  else:
    entrada_clean = int(entrada)
    if entrada_clean == 0:
      break
    int_list.append(entrada_clean)

# CALCULO
for i in range(len(int_list)):
  if int_list[i] > 0:
    soma_pos += int_list[i]
  elif int_list[i] < 0:
    continue

for i in range(len(int_list)):
  if int_list[i] % 2 == 0:
    cont_par += 1
  elif int_list[i] % 2 != 0:
    cont_impar += 1

media = sum(int_list) / len(int_list)

print(linha)
print(f'Lista inicial: {int_list}')
print(linha)
print(f'Soma dos números positivos: {soma_pos}')
print(f'Quantidade de números pares: {cont_par}')
print(f'Quantidade de números ímpares: {cont_impar}')
print(f'Média dos números: {media}')
print(linha)
for i in range(len(int_list) -1, -1, -1):
  print(int_list[i])
```
