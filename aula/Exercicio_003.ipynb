# Exercício 3 - Formatação de Saída com sep e end

**Enunciado:**
Utilizando os parâmetros sep (separador) e end (finalizador) da função print(),
elabore um programa que exiba uma lista de itens em uma única linha, com os elementos
separados por um caractere definido pelo programador.

**Orientações:**
* Crie uma lista contendo, no mínimo, cinco elementos;
* Utilize sep para definir o separador entre os elementos;
* Uilize end para impedir a quebra de linha ao final da impressão ou para personalizá-la.

```python
lista = ['pera', 'uva', 'maça', 3, 2, 1]

print(*lista, sep = ' + ', end=' = fim da lista.')
```
