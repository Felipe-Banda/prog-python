# Exercício 6 - Sisitema de Matriz com Bubble Sort e Operações Bit a Bit

**Enunciado:**
Crie um sistema completo que trabalhe com uma matriz 3x3 (lista de listas) e combine vários conceitos:
* Crie uma matriz 3x3 com valores digitados pelo usuário (loops aninhados for);
* Imprima a matriz formatada (cada linha em uma linha separada);
* Calcule e imprima:
  * A soma de cada linha;
  * A soma de cada coluna;
  * A diagonal principal (elementos onde linha == coluna).
* Aplique o bubble sort para ordenar todos os elementos da matriz em uma única lista, depois reorganize a matriz ordenada;
* Use operações bit a bit para:
  * Verificar se um número é par ou ímpar usando & 1 (se num & 1 == 0, é par);
  * Dobrar um número usando deslocamento << 1;
  * Dividir por 2 usando deslocamento >> 1.
* Use in / not in para verificar se um número procurado existe na matriz.

```python
# Inicializações
ln = '=' * 22
col, lin = 3, 3
soma_lin = [0] * lin
soma_col = [0] * col
diag_prin = [0] * lin

# I/O matriz
table = [[int(input(f'Digite um valor para ({j}, {i}): ')) for i in range(col)] for j in range(lin)]
print(ln)
print('Matriz do usuário:', *table, sep = '\n')
print(ln)

# Processamento
for i in range(lin):
  for j in range(col):
    soma_lin[i] += table[i][j]
    soma_col[j] += table[i][j]
    if i == j:
      diag_prin[i] = table[i][j]

# Desempacotamento da matriz
table_sorted = []
for i in range(lin):
  for j in range(col):
    table_sorted.append(table[i][j])

# Bubble sort
monitor = True
while monitor:
  monitor = False
  for i in range(len(table_sorted) - 1):
    if table_sorted[i] > table_sorted[i + 1]:
      temp = table_sorted[i]
      table_sorted[i] = table_sorted[i + 1]
      table_sorted[i + 1] = temp
      monitor = True

# Matriz ordenada
for i in range(lin):
  for j in range(col):
    table[i][j] = table_sorted[i * col + j]
    # Verificação bit a bit
    if table[i][j] & 1 == 0:
      print(f'O número {table[i][j]} é par.')
    else:
      print(f'O número {table[i][j]} é ímpar.')
    if table[i][j] << 1 in table:
      print(f'{table[i][j]} << 1 = {table[i][j] << 1} (existe na matriz).')
    elif table[i][j] << 1 not in table:
      print(f'{table[i][j]} << 1 = {table[i][j] << 1} (não existe na matriz).')
    if table[i][j] >> 1 in table:
      print(f'{table[i][j]} >> 1 = {table[i][j] >> 1} (existe na matriz).')
    elif table[i][j] >> 1 not in table:
      print(f'{table[i][j]} >> 1 = {table[i][j] >> 1} (não existe na matriz).')
    print(ln)

print(ln)
print(f'A soma de cada linha é: {soma_lin}')
print(f'A soma de cada coluna é: {soma_col}')
print(f'Valores da diagonal principal: {diag_prin}')
print(ln)
print('Matriz ordenda:', *table, sep = '\n')
print(ln)
```
