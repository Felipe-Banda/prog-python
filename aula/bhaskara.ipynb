```python
# Entrada de Dados
a = float(input("Insira o valor de a: "))
b = float(input("Insira o valor de b: "))
c = float(input("Insira o valor de c: "))

print("Equação definida:", a,"x² +", b,"x +", c,"= 0")

# Processamento + Saída
delta = round(b**2 - 4*a*c, 2)
print("Delta =", delta)
if(a == 0):
    print("ERRO: 'a' não pode ser 0")
else:
  if(delta > 0):
    x1 = round(((-b + delta**0.5) / (2*a)), 2)
    x2 = round(((-b - delta**0.5) / (2*a)), 2)
    print("x1 =", x1)
    print("x2 =", x2)
  elif(delta == 0):
    x = round(-b / (2*a))
    print("x =", x)
  elif(delta < 0):
    print("Nenhuma raíz real")
```
